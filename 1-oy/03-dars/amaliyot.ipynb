## 4. Tozalash: bo’sh qiymat, takror, noto’g’ri tur

Real ma’lumot deyarli hech qachon toza bo’lmaydi. Uch asosiy muammo:
- **bo’sh qiymat** (`NaN`) — ma’lumot yozilmagan;
- **takror** — bir qator ikki marta kiritilgan;
- **noto’g’ri tur** — son matn bo’lib qolgan (`"14"` o’rniga `14` kerak).

Avval Titanic'da bo’sh qiymatlarni sanaymiz.

```python
print(df.isna().sum()[df.isna().sum() > 0])        # faqat bo'sh qiymati bor ustunlar
print((df.isna().mean() * 100).round(1)["deck"])   # deck ustunining necha foizi bo'sh
```

`deck` ning 77% i bo’sh — bunday ustunni to’ldirib bo’lmaydi, uni tashlab yuborish to’g’riroq. `age` ning 20% i bo’sh — buni to’ldirish mumkin. Uch yo’l bor:

1. **Qatorni o’chirish** (`dropna`) — 177 yo’lovchidan ayriladi, ma’lumot kamayadi.
2. **Umumiy median bilan to’ldirish** — oddiy, lekin hammaga bir xil yosh beradi.
3. **Guruh mediani bilan to’ldirish** — har sinf uchun o’z yoshi.

Qaysi biri yaxshi? Sinf bo’yicha yoshlarni ko’ramiz.

```python
print("Umumiy median:", df["age"].median())
print(df.groupby("pclass")["age"].median())        # sinf bo'yicha median yosh
```

## 5. Vizualizatsiya: ma’lumotni ko’z bilan ko’rish

Raqamlar jadvalida naqsh ko’rinmaydi, grafikda esa bir qarashda ko’rinadi. Uchta kutubxona birga ishlaydi:

Grafik turini savolga qarab tanlanadi:

| Savol | Grafik | seaborn |
|---|---|---|

## 7. Noxolislik (bias): ma’lumot haqiqatni to’liq aks ettiradimi?

Model faqat unga ko’rsatilgan ma’lumotdan o’rganadi. Ma’lumotda nuqson bo’lsa, model ham o’sha nuqsonni o’rganadi. Ikkita misol.

**1. Bo’sh qiymatlar tasodifiy emas.** Yoshi yozilmaganlar omon qolganlarmi yoki halok bo’lganlarmi?

```python
bosh_yosh = df["age"].isna()                              # True — yoshi yozilmagan
print(df.groupby(bosh_yosh)["survived"].mean())           # ikki guruhda omon qolish ulushi
print(df.groupby("pclass")["age"].apply(lambda s: s.isna().mean()))   # sinf bo'yicha bo'sh ulushi
```

## Chuqurroq (ixtiyoriy)

`pivot_table` — ikki o’lchamli jadval: qatorlarda bir ustun, ustunlarda boshqasi, ichida o’rtacha. Uni `heatmap` bilan chizsa, jins va sinf birgalikdagi ta’siri bir qarashda ko’rinadi.

```python
jadval = df.pivot_table(index="pclass", columns="sex", values="survived", aggfunc="mean")
print(jadval)
sns.heatmap(jadval, annot=True, cmap="Greens", vmin=0, vmax=1)
plt.title("Sinf va jins bo'yicha omon qolish ulushi")
plt.show()
```
