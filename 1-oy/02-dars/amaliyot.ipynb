## 0. Tayyorgarlik: 1-darsdagi raqam tanuvchi

Bu katak 1-darsdagi hamma narsani qayta yasaydi: rasmlar, train/test va sof Python’dagi 1-NN. Oxirida vaqtni o’lchaydi — bugun shu vaqtni yengamiz.

```python
from sklearn.datasets import load_digits
import random, math, time

raqamlar = load_digits()
X = raqamlar.data.tolist()
y = raqamlar.target.tolist()

juftlar = list(zip(X, y))                    # 1-darsdagi train/test
random.Random(42).shuffle(juftlar)
test, train = juftlar[:300], juftlar[300:]
X_train = [x for x, _ in train]
y_train = [t for _, t in train]
X_test = [x for x, _ in test]
y_test = [t for _, t in test]


def masofa(a, b):                            # 1-darsdagi masofa — sikl bilan
    return math.sqrt(sum((p - q) ** 2 for p, q in zip(a, b)))


def bashorat_1nn(rasm, X, y):                # 1-darsdagi 1-NN — sikl bilan
    eng_kichik, javob = float("inf"), None
    for x, t in zip(X, y):
        m = masofa(rasm, x)
        if m < eng_kichik:
            eng_kichik, javob = m, t
    return javob


bosh = time.time()                           # vaqtni o'lchash: boshlanish
bashoratlar = [bashorat_1nn(x, X_train, y_train) for x in X_test]
vaqt_python = time.time() - bosh             # tugash - boshlanish = sarflangan vaqt
togri = sum(b == t for b, t in zip(bashoratlar, y_test))
print(f"Sof Python: {100 * togri / len(y_test):.1f}%, {vaqt_python:.1f} soniya")
```

## 4. Matritsa ko’paytirish: ko’p dot product bir yo’la

Sinfdoshlarni bitta jadvalga yig’amiz: har qator — bitta odam. `S @ men` — `S` ning **har qatori** bilan dot product, bir yo’la.

Shakllar qoidasi: `(n, k) @ (k,)` → `(n,)`. Chapning ustunlari soni o’ngning uzunligiga teng bo’lishi kerak. Ikki jadval uchun: `(n, k) @ (k, m)` → `(n, m)`.

```python
ismlar = list(sinfdoshlar)
S = np.array(list(sinfdoshlar.values()))   # 5 qator (odam) × 5 ustun (qiziqish)
print(S.shape, "@", men.shape)

ballar = S @ men                            # 5 ta dot product — bitta amal
print(ballar)

oxshash = ballar / (np.linalg.norm(S, axis=1) * np.linalg.norm(men))   # hammasi uchun kosinus
print(np.round(oxshash, 2))
print("Menga eng o'xshash:", ismlar[oxshash.argmax()])
```
